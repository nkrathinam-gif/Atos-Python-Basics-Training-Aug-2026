```python
#   Ticket Priority Classification
priority = "P1"
match priority:
    case "P1":
        action = "Immediate Response"
    case "P2":
        action = "Respond within four hours"
    case "P3":
        action = "Respond within twenty-four hours"
    case _:                             #   Default Case
        action = "Standard Queue"
print(action)
```

```python
#   Even / Odd Checker
num = int(input("Enter a number:   "))
is_even = (num %2 == 0)
print("Even? ", is_even)
print(f"Is the number even?{is_even}")          # f stands for format string
```

```python
#   Even / Odd Checker
num = 1
while num > 0:
    num = int(input("Enter a number [ 0 or -ve to stop]:   "))
    if num <=0:
        break
    is_even = (num %2 == 0)
    print("Even? ", is_even)
    print(f"Is the number even?{is_even}")          # f stands for format string
print("As the number is zero or lesser than zero, the code breaks")
print()
```

```python
#   Even / Odd Checker using While True - infinite loop, which should break with break statement
while True:                                                                                          #   When we set True, the loop become infinite
    num = int(input("Enter a number [ 0 or -ve to stop]:   "))
    if num <=0:
        break
    is_even = (num %2 == 0)
    print("Even? ", is_even)
    print(f"Is the number even?{is_even}")                                      #   f stands for format string
print("As the number is zero or lesser than zero, the code breaks")
print()
```

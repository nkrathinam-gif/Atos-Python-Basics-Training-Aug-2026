```python
# 13. Reverse a list without using built in functions.
my_list = ['a','e','i','o','u']
print(f"My List = {my_list}")
l1 = len(my_list) - 1
rev_list = []
while l1 >= 0:
  rev_list.append(my_list[l1])
  l1 = l1 - 1
print(f"Reversed List = {rev_list}")
```

```python
#   03. Employee Bonus Calculator
#   Scenario: Calculate bonus based on salary and performance.
#   Tasks:
#     •	Input salary and rating.
#     •	Apply bonus rules.
#     •	Display final salary.
print()
cursal = int(input("Enter your current salary:    "))
rating = int(input("Enter your performance rating [1 to 5]:   "))
if rating == 1:
  newsal = cursal + (1+(0.025/100))
  print(f"Your new salary based on your current salary and rating is: {newsal}")
elif rating == 2:
  newsal = cursal + (1+(0.05/100))
  print(f"Your new salary based on your current salary and rating is: {newsal}")
elif rating == 3:
  newsal = cursal + (1+(0.07/100))
  print(f"Your new salary based on your current salary and rating is: {newsal}")
elif rating == 4:
  newsal = cursal + (1+(0.09/100))
  print(f"Your new salary based on your current salary and rating is: {newsal}")
elif rating == 5:
  newsal = cursal + (1+(0.1/100))
  print(f"Your new salary based on your current salary and rating is: {newsal}")
else:
  print("Please enter a rating value within 1 to 5")
```

```python
#   04. Password Validator
#   Scenario: Basic password strength checker.
#   Tasks:
#     •	Input password.
#     •	Check length and numeric presence.
#     •	Display validity.
print()

#  Prereq
my_integers = list(range(0,10))
my_alphabets = list(map(chr, range(97,123)))
my_pass_len_result = False
my_pass_num_result = False

#  Get the new password
my_password = input("Enter your password [8 to 16 char length with alphabet and numbers]:    ")
my_pass_list = list(my_password)    # Convert the password into a list
my_pass_len = len(my_password)      # Get the length of the password

# Check the length of the password
if my_pass_len >= 8 and my_pass_len < 16:
  my_pass_len_result = True
  print("The password is of required length")
else:
  my_pass_len_result = False
  print("The password doesn't satisify the password length standards")

# Check for the numeric presence
for i in my_pass_list:
  if i.isdigit():
    my_pass_num_result = True
    print("The password has numeric value")
  else:
      my_pass_num_result = False
      print("The password doesn't have a numeric value")

# Final check whether password meets both length and numeric requirements
if my_pass_len_result == True and my_pass_num_result == True:
  print("The New Password is Compliant!")
else:
  print("The Password is non-compliant, please try again!")
```

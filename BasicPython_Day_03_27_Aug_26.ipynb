```python
#   Display Even or Odd  Numbers
for num in range(1,21):
    if num % 2 == 0:
        print("This is an even number:    ",num)
    else:
        if num % 2!= 0:
            print("This is an Odd  number:    ",num)
        else:
            continue
```

```python
#   Monthly Sales Analysis
sales = [4500, 5000, 3000, 5000, 6000, 3000 ]
target = 3500
for month, amount in enumerate(sales, start=1):
    if amount >= target:
        print(f"Month {month}:   Target Achieved!")
    else:
        print(f"Month {month}:   Target Missed!")
```

```python
#   Training Attendance Monitoring
attendance = [ 95,82,68,91,55]
for empno, percent in enumerate(attendance, start = 1):
    if percent < 75:
        print(f"Employee  {empno}:   Improvement Required")
        continue
    print(f"Employee {empno}:   Compliant!")
```

```python
#   Find Largest Sales
sales = [ 45000, 52000, 38000, 61000, 72000, 20000 ]
largest = sales[0]
for amount in sales:
    if amount > largest:
        largest = amount
print(largest)
```

```python
#   Nested Loops
for i in range(1,6):
    for j in range(1, 6):
        print( i * j, end = "\t")
    print()
```

```python
#   Show attendance across batches and participants
batches = ["Batch A", "Batch B"]
participants = ["John", "Mary", "Sam"]
for batch in batches:
    print(batch)
    for person in participants:
        print(" ", person)
```

```python
#   Find Duplicate Values
numbers = [1,2,3,4,2,5,3]
seen = set()
for num in numbers:
    if num in seen:
        print("Duplicate:   ", num)
    seen.add(num)
print(seen)
```

```python
#   Prime Number Check
number = 17
is_prime = True
for i in range(2,number):
    if number % i == 0:
        is_prime = False
        break
if is_prime:
    print("Prime Number")
else:
    print("Not Prime")
```

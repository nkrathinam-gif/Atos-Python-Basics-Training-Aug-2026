```python
#       Syntax for input is: input("what do you want the user to input")
name = input("What is your name:    ")
print("Hello "+name)
#       Combined code of print and name
print("Hello " + input("Whats is your name")+" !!!")
```

```python
#       Assign variable with the user inputs
#       When we just use input, we don't know where the value goes which is provided by user
#       So use variables to capture the data, so that you can manipulate, operate or print it
#       In the below code, the "name" is known as variable which holds the value
#       To print a vairable, don't provide it within single or double quotes
#       When we provide a single or double quotes whatever within it gets printed
#       Within a code, whatever is the last assigned value to the variable will be remembered
name = input("What is your name?")
print(name)
name = "Jack"
print(name)
```

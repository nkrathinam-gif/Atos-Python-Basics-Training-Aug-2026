```python
# Conditional Logic - if statement
print('choose your drink!')
print('1. Tea')
print('2. Coffee')
print('3. Soft Dink')
print('4. Beer')
choice = int(input("What's your choice? "))
print(choice)
if choice == 1:
  print("You like Tea!")
elif choice == 2:
  print("You like Coffee!")
elif choice == 3:
  print("You like Soft Drink!")
elif choice == 4:
  print("You like Beer!")
else:
  print("Invalid choice")
```

```python
import sys
print('Choose your drink!')
print('1. Tea')
print('2. Coffee')
print('3. Soft Drink')
print('4. Beer')
choice = int(input("What's your choice? "))
match choice:
  case 1:
   print("You like Tea!")
  case 2:
   print("You like Coffee!")
  case 3:
   print("You like Soft Drink!")
  case 4:
   print("You like Beer!")
  case _:
   print("Wrong Choice!")
```

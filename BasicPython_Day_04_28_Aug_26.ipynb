```python
#   01
#   Python Data Structures
#   List - Ordered, Mutable (changeable), Allows Duplicates
#   while doing the project, the lists can be updated, ammended and replaced
#   To define a List of fruits
#   Lists are defined with square backets [ ]
fruits = [ "Apple", "Bannana", "Mango", "Grapes"]       #   Ensure that the string values are within double quotes , but single quotes can also be used
print(fruits)
#   To append the existing list
fruits.append("Cherry")
print(fruits)
#   To remove an element from the existing list
fruits.remove("Cherry")
print(fruits)
#   Access Specific element
print("First Element in the Fruits List:", fruits[0])
#   List Slicing
num = [ 10, 20, 30, 40, 50 ]
print(num)
#   To print the first n elements in the list
print("First 3 Elements in the Fruits List are:", num[:3])
#   To print the last n elements in the list
print("Last 3 Elements in the Fruits List are:", num[-2:])
#   To print the middle of  n elements in the list
print("Middle 3 Elements in the Fruits List are:", num[1:4])
#   To update an element
num[2] = 70
print(num)
```

```python
#   02
#   Tuple (Ordered, Immutable (not changeable))
#   cord contains the coordinates values
#   Tuples are defined with paranthesis ()
cord = (10,20)
print("Coordinates Values:   ", cord)
print("X Coordinates from cord:   ", cord[0])
print("Y Coordinates from cord:   ", cord[1])
#   Unlike List the Tuples are immutable
#   So you can't update the Tuple
#   This would not be executed
#   The below code is commented because it would not be executed
#   cord[1] = 100
#   But if we need to update, the whole thing can be updated or changed
#   Only partial update of the Tuples are not allowed
cord = (70,80)
print(cord)
```

```python
#   03
#   Python Data Structures
#   List - Ordered, Mutable (changeable), Allows Duplicates
#   while doing the project, the lists can be updated, ammended and replaced
#   To define a List of fruits
fruits = [ "Apple", "Bannana", "Mango", "Grapes"]       #   Ensure that the string values are within double quotes , but single quotes can also be used
print(fruits)
size = len(fruits)
print(size)
i = 0
while i <= size:
    print(fruits[i])
    i = i+1
break
```

```python
#   04
#   Set (Unordered, Mutable, No Duplicates)
#   sets are defined with curly bracets {}
nums = { 1, 2, 3, 3, 4 }
print(nums)
#   In List, addition of elements is through append
#   In sets, addition of elements is through add
nums.add(5)
print(nums)
```

```python
#   05
#   Dictionary (key-value pairs)
student = {
    "name": "Shilpa",
    "gender": "Female",
    "age": 32,
    "class": "X"
}
print(student)
print("Name of student", student["name"])
print("Class of student",student["class"])
student["class"] = "XI"
student["subject"] = "Maths"
print(student)
```

```python
#   06
#   Looping through data structure
#   Loop through List
for fruit in fruits:
    print(fruit)
#   Loop through Dictionary
for key, value in student.items():
    print(key, ":", value)
```

```python
#   07
#   Nested Data Structure
#   Students have multiple information of the students
student_names = [ 'Vijay', 'Suhas', 'Ameya', 'Preeti']
student_marks = [ 50, 60, 70, 40 ]
students = [
    {"Name": "Amit", "Marks": 85},
    {"Name": "Neha", "Marks": 90}
]
print(student_names)
print("-"*30)
print(student_marks)
print("-"*30)
print(students)
print("-"*30)
for s in students:
    print(s)
print("-"*30)
for s in students:
    print(s["Name"], "Scored", s["Marks"])
print("-"*30)
```

```python
#   08
#   Common Operations on the data structures
#   Define List
age = [ 10, 10, 11, 12, 13, 15 ]
#   To find the length of the list, the statement is "len"
print("Length of the list:   ", len(age))
#   To know whether the element is present in the List, operator is "in"
print("Does the group have someone with age 12?   ", 12 in age)
#   Sorting
age.sort()
print("Sorted Age List is    ",age)
```

```python
#   09
#   Set Operations
a = { 1, 2, 3 }
b = { 3, 4, 5 }
print("Union of a and b:   ", a|b)
print("Intersection of a and b:    ",a&b )
print("Difference:   ", a-b)
print("Difference:   ", b-a)
```

```python
#   10
#   Dictionary Methods
student = {
    "name": "Shilpa",
    "gender": "Female",
    "age": 32,
    "class": "X"
}
print("Keys:   ", student.keys())
print("Values:   ", student.values())
print("Items:   ", student.items())
```

```python
#   11
#   Shopping Cart Example
cart = []
#   Add Items t to the cart
cart.append({"Item" : "Laptop", "Price" : 50000})
cart.append({"Item" : "Keyboard", "Price" : 1500})
cart.append({"Item" : "Mouse", "Price" : 500})
print(cart)
#   To Calculate the total cost
total = 0
for product in cart:
    total += product["Price"]
print("Cart Contains:   ", len(cart), " items")
print("Total Amount to be paid: ", total)
```

```python
#   User Defined Function ( UDF )
#   You can define the UDF once and then call this program whenever / wherever we require
def greet():
    print("Hello! Welcome to Python!")
greet()
print("Tested UDF")
print("Call Again")
greet()
```

```python
#   Function with parameters and return value
def add(a,b):
    return a+b
result = add(2,1)
print(result)
```

```python
# Default Parameters
def power1(base,exp):
    return base ** exp
result = power1(2,3)
print(result)
def power2(base,exp=2):
    return base ** exp
print("Square:   ", power2(2))
print("Cube:   ", power1(5,3))
```

```python
#   Multiple Return Values
def calculate(a,b):
    return a+b, a-b, a*b, a/b, a**b
a, s, m, d, p = calculate(10,5)
print("Sum is   ", a, "Subtract is   ", s, "Multiplication is   ", m, "Division is   ",d, "Power is   ",p)
```

```python
#   Variable Length Arguments ( *args)
def total_sum(*numbers):
    return sum(numbers)
print(total_sum(1,2))
print(total_sum(1,2,3,4,5,6,7))
```

```python
#   Keywords arguments (**kwargs - Key word arugments)
def print_details(**info):
    for key, value in info.items():
        print(f"{key}:{value}")

print_details(name = "Vijay", role = "Trainer", dept = "HR", city = "Pune")
print()
print_details(id = "101", desc  = "Biscuits", price = 30)
```
